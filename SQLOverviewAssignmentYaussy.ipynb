```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("chinook.db")
```

```python
tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

display(tables)
```

```python
query1 = '''
SELECT FirstName, LastName, Country
FROM customers
WHERE Country = 'Canada'
ORDER BY LastName ASC;
'''

display(pd.read_sql_query(query1, conn))
```

```python
query2 = '''
SELECT Name, Milliseconds
FROM tracks
ORDER BY Milliseconds DESC
LIMIT 10;
'''

display(pd.read_sql_query(query2, conn))
```

```python
query3 = '''
SELECT artists.Name AS Artist,
       albums.Title AS Album
FROM artists
INNER JOIN albums
    ON artists.ArtistId = albums.ArtistId
WHERE artists.Name = "AC/DC";
'''

display(pd.read_sql_query(query3, conn))
```

```python
query4 = '''
SELECT tracks.Name AS Song,
       genres.Name AS Genre,
       tracks.UnitPrice
FROM tracks
INNER JOIN genres
    ON tracks.GenreId = genres.GenreId
WHERE genres.Name = "Jazz"
ORDER BY tracks.Name ASC;
'''

display(pd.read_sql_query(query4, conn))
```

```python
query5 = '''
SELECT artists.Name AS Artist
FROM artists
LEFT JOIN albums
    ON artists.ArtistId = albums.ArtistId
WHERE albums.AlbumId IS NULL
ORDER BY artists.Name ASC;
'''

display(pd.read_sql_query(query5, conn))
```
